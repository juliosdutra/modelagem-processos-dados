```python
# 1. Importando as bibliotecas
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2. Definindo os dados
TOC = [21, 18, 43, 21, 11, 18, 22, 19, 24, 25, 15, 20, 26, 23, 22, 18, 27]

# Simulando a presença de um outlier
TOC2 = [2000, 21, 18, 43, 21, 11, 18, 22, 19, 24, 25, 15, 20, 26, 23, 22, 18, 27]


plt.plot(TOC, 'o', label='Dados')
plt.axhline(np.mean(TOC), color='r', linestyle='--', label='Média')
plt.axhline(np.percentile(TOC,50), color='green', linestyle='-.', label='Mediana')
plt.xlabel('Amostra')
plt.ylabel('TOC [ppm]')
plt.legend()
plt.show()
```

```python
def analise_estatistica(dados):
  x_media = np.mean(dados)
  x_mediana = np.percentile(dados,50)
  x_sd = np.std(dados)
  x_IQR = np.percentile(dados,75) - np.percentile(dados,25)
  print(f'Média: {x_media}')
  print(f'Mediana: {x_mediana}')
  print(f'Desvio padrão: {x_sd}')
  print(f'Intervalo interquartil: {x_IQR}')
```

```python
for dados in [TOC, TOC2]:
  analise_estatistica(dados)
  print(' ')
```

```python
plt.figure(figsize=(6, 4))
plt.boxplot(TOC, vert=False, tick_labels=['TOC'],
            patch_artist=True, boxprops=dict(facecolor="lightblue", color="blue"),
            medianprops=dict(color="red", linewidth=2),
            whiskerprops=dict(color="blue"),
            capprops=dict(color="blue"),
            flierprops=dict(marker="o", color="darkred", alpha=0.6))

plt.title("Boxplot dos valores de TOC (ppm)")
plt.xlabel("TOC (ppm)")
plt.grid(True, axis="x", linestyle="--", alpha=0.6)
plt.show()
```
