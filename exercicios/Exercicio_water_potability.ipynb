```python
# --- Leitura e análise inicial do dataset water_potability.csv ---

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# URL do arquivo no GitHub (raw)
url = "https://raw.githubusercontent.com/juliosdutra/modelagem-processos-dados/refs/heads/main/dados/water_potability.csv"

# Ler o dataset
df = pd.read_csv(url)

# Mostrar dimensões e primeiras linhas
print("Formato do dataset:", df.shape)
df.head()
```

```python
# Informações gerais
df.info()

# Estatísticas descritivas
df.describe()
```

```python
import missingno as msno

print("Valores ausentes por coluna:")
print(df.isnull().sum())

# Visualização dos valores ausentes
msno.matrix(df)
plt.show()
```

```python
# Histograma para algumas variáveis
df.hist(figsize=(12, 8), bins=30)
plt.suptitle("Distribuições das variáveis", fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), cmap="coolwarm", annot=False)
plt.title("Matriz de correlação")
plt.show()
```
