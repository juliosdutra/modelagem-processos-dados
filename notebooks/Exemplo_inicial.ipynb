```python
# Exemplo inicial - PPEQ-1040
# "Hello World" da Modelagem Baseada em Dados

import numpy as np
import matplotlib.pyplot as plt

# Dados sintéticos
x = np.linspace(0, 10, 50)
y = 2.5 * x + 3 + np.random.normal(0, 2, 50)

# Ajuste de regressão linear simples
coef = np.polyfit(x, y, 1)
y_pred = np.polyval(coef, x)

plt.scatter(x, y, label="Dados reais", color="blue")
plt.plot(x, y_pred, label="Modelo ajustado", color="red")
plt.legend()
plt.title("Exemplo inicial - Regressão Linear")
plt.show()

print("Coeficientes ajustados:", coef)
```
